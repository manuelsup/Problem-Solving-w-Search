```python
# --- Puzzle Setup ---
import random
import heapq
from collections import deque
import time
import pandas as pd


def print_puzzle(state, size):
    for i in range(size):
        print(' '.join(f"{n:2}" if n != 0 else "  " for n in state[i*size:(i+1)*size]))
    print()

def goal_state(size):
    return tuple(range(1, size*size)) + (0,)

def get_neighbors(state, size):
    neighbors = []
    zero = state.index(0)
    row, col = divmod(zero, size)
    moves = []
    if row > 0: moves.append((-1, 0))
    if row < size - 1: moves.append((1, 0))
    if col > 0: moves.append((0, -1))
    if col < size - 1: moves.append((0, 1))

    for dr, dc in moves:
        new_row, new_col = row + dr, col + dc
        new_zero = new_row * size + new_col
        new_state = list(state)
        new_state[zero], new_state[new_zero] = new_state[new_zero], new_state[zero]
        neighbors.append(tuple(new_state))
    return neighbors

def random_walk(start, size, steps):
    state = start
    for _ in range(steps):
        state = random.choice(get_neighbors(state, size))
    return state
```

```python
# --- Heuristics ---
def misplaced_tiles(state, goal, size):
    return sum(1 for i in range(len(state)) if state[i] != 0 and state[i] != goal[i])

def manhattan_distance(state, goal, size):
    dist = 0
    for idx, tile in enumerate(state):
        if tile == 0: continue
        goal_idx = goal.index(tile)
        x1, y1 = divmod(idx, size)
        x2, y2 = divmod(goal_idx, size)
        dist += abs(x1 - x2) + abs(y1 - y2)
    return dist
```

```python
# --- Search Algorithms ---
def bfs(start, goal, size, max_nodes=10000, max_time=5):
    start_time = time.time()
    frontier = deque([[start]])
    visited = set()
    nodes_expanded = 0
    while frontier:
        if nodes_expanded > max_nodes or (time.time() - start_time) > max_time:
            return [], nodes_expanded
        path = frontier.popleft()
        state = path[-1]
        if state in visited:
            continue
        visited.add(state)
        nodes_expanded += 1
        if state == goal:
            return path, nodes_expanded
        for neighbor in get_neighbors(state, size):
            if neighbor not in visited:
                frontier.append(path + [neighbor])
    return [], nodes_expanded

def astar(start, goal, size, heuristic, max_nodes=10000, max_time=5):
    start_time = time.time()
    frontier = [(heuristic(start, goal, size), 0, [start])]
    visited = set()
    nodes_expanded = 0
    while frontier:
        if nodes_expanded > max_nodes or (time.time() - start_time) > max_time:
            return [], nodes_expanded
        est_total, cost, path = heapq.heappop(frontier)
        state = path[-1]
        if state in visited:
            continue
        visited.add(state)
        nodes_expanded += 1
        if state == goal:
            return path, nodes_expanded
        for neighbor in get_neighbors(state, size):
            if neighbor not in visited:
                g = cost + 1
                h = heuristic(neighbor, goal, size)
                heapq.heappush(frontier, (g + h, g, path + [neighbor]))
    return [], nodes_expanded
```

```python
# --- Experiment Runner ---
def run_experiments():
    results = []
    for size in [3, 4]:
        goal = goal_state(size)
        for steps in [5, 10, 20, 40, 80]:
            for _ in range(3):
                start = random_walk(goal, size, steps)
                row = {'size': size, 'steps': steps, 'start': start}
                bfs_path, bfs_nodes = bfs(start, goal, size)
                row['bfs_len'] = len(bfs_path) - 1
                row['bfs_nodes'] = bfs_nodes
                a1_path, a1_nodes = astar(start, goal, size, misplaced_tiles)
                row['a*_misplaced_len'] = len(a1_path) - 1
                row['a*_misplaced_nodes'] = a1_nodes
                a2_path, a2_nodes = astar(start, goal, size, manhattan_distance)
                row['a*_manhattan_len'] = len(a2_path) - 1
                row['a*_manhattan_nodes'] = a2_nodes
                results.append(row)
    df = pd.DataFrame(results)
    print(df.to_string(index=False))
```

```python
run_experiments()
```

These results show that as problem size increases from 3×3 to 4×4, uninformed search like BFS becomes inefficient, often running out of time or memory. However, informed search using A* with good heuristics (like Manhattan distance) performs much better by guiding the search toward the goal. This highlights how heuristic-based search is essential in AI for solving complex problems efficiently.
